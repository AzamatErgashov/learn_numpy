# Looplarning sekinligi.

Pythonning standart ilovasi (CPython sifatida tanilgan) ba'zi operatsiyalarni juda sekin bajaradi. Bu qisman tilning dinamik, talqin qilinadigan tabiati bilan bog'liq: turlarning moslashuvchanligi, shuning uchun operatsiyalar ketma-ketligini C va Fortran kabi tillarda bo'lgani kabi samarali mashina kodiga yig'ib bo'lmaydi. So'nggi paytlarda ushbu zaiflikni bartaraf etish uchun turli xil urinishlar bo'ldi: taniqli misollar - Python dasturining o'z vaqtida tuzilgan PyPy loyihasi; Python kodini kompilyatsiya qilinadigan C kodiga aylantiruvchi Cython loyihasi; va Python kodining parchalarini tez LLVM baytekodiga aylantiruvchi Numba loyihasi. Ularning har biri o'zining kuchli va zaif tomonlariga ega, ammo ishonch bilan aytish mumkinki, uchta yondashuvning hech biri hali standart CPython dvigatelining qo'llanilishi va mashhurligidan oshib ketmagan.
Python-ning nisbatan sustligi odatda ko'plab kichik operatsiyalar takrorlanadigan holatlarda o'zini namoyon qiladi, masalan, har bir elementda ishlash uchun massivlar ustida aylanish. Misol uchun, bizda bir qator qiymatlar borligini tasavvur qiling va biz har birining o'zaro nisbatini hisoblamoqchimiz. To'g'ridan-to'g'ri yondashuv shunday ko'rinishi mumkin:

```python
import numpy as np  
np.random.seed(0)
def compute_reciprocals(values):  
    output = np.empty(len(values))   
    for i in range(len(values)):  
        output[i] = 1.0 / values[i]    
        return output
values = np.random.randint(1, 10, size=5)   
compute_reciprocals(values)
```

Ushbu dastur, masalan, C yoki Java fonida bo'lgan kishi uchun juda tabiiy tuyulishi mumkin. Ammo agar biz ushbu kodning bajarilish vaqtini katta kirish uchun o'lchaydigan bo'lsak, bu operatsiya juda sekin ekanligini ko'ramiz, ehtimol hayratlanarli! Biz buni IPython-ning %timeit sehri bilan taqqoslaymiz (25-betdagi “Profillash va vaqt kodi”da muhokama qilingan):

```python
big_array = np.random.randint(1, 100, size=1000000)
%timeit compute_reciprocals(big_array)
```

funktsiyasi:


```python
import numpy as np
np.empty(10)
```

```python
x = np.arange(5)
y = np.empty(5)                       
np.multiply(x, 10, out=y)
print(y)
```

Bu hatto massiv ko'rinishlari bilan ham ishlatilishi mumkin. Masalan, biz hisoblash natijalarini belgilangan massivning har bir boshqa elementiga yozishimiz mumkin:

Agar biz o'rniga y[::2] = 2 ** x yozganimizda, bu 2 ** x natijalarini saqlash uchun vaqtinchalik massivni yaratishga olib kelgan bo'lar edi, keyin esa bu qiymatlarni y massiviga ko'chiradigan ikkinchi operatsiya. . ## Oldingi bo'limda biz NumPy-ning universal funktsiyalaridan operatsiyalarni vektorlashtirish va shu bilan sekin Python looplarini olib tashlash uchun qanday foydalanish mumkinligini ko'rib chiqdik. Operatsiyalarni vektorlashtirishning yana bir usuli - NumPy-ning eshittirish funksiyasidan foydalanish. Eshittirish oddiygina turli o'lchamdagi massivlarda ikkilik ufunklarni (qo'shish, ayirish, ko'paytirish va hokazo) qo'llash qoidalari to'plamidir..


# Radioeshittirish bilan tanishtirish


# Eslatib o'tamiz, bir xil o'lchamdagi massivlar uchun ikkilik operatsiyalar element bo'yicha bajariladi.:


```python
import numpy as np
a = np.array([0, 1, 2])
b = np.array([5, 5, 5])
a + b
```

Radioeshittirish bu turdagi ikkilik operatsiyalarni turli o'lchamdagi massivlarda bajarishga imkon beradi - masalan, biz massivga skalerni (uni nol o'lchovli massiv deb o'ylab ko'ring) osongina qo'shishimiz mumkin:


```python
a + 5
```

Buni 5 qiymatini massivga [5, 5, 5] cho‘zuvchi yoki ko‘paytiruvchi va natijalarni qo‘shuvchi operatsiya deb tasavvur qilishimiz mumkin. NumPy-ning efirga uzatilishining afzalligi shundaki, qadriyatlarning bu takrorlanishi aslida sodir bo'lmaydi, lekin biz eshittirish haqida o'ylaganimizda, bu foydali aqliy modeldir.
Xuddi shunday, biz buni yuqori o'lchamdagi massivlarga ham kengaytirishimiz mumkin. Ikki o'lchovli massivga bir o'lchovli massiv qo'shganda natijani kuzating:


```python
M = np.ones((3, 3))
M
```

```python
M + a
```

## Bu erda bir o'lchovli massiv M shakliga mos kelish uchun ikkinchi o'lchov bo'ylab uzatiladi yoki uzatiladi.
## Ushbu misollarni tushunish nisbatan oson bo'lsa-da, murakkabroq holatlar ikkala massivni ham translyatsiya qilishni o'z ichiga olishi mumkin. Quyidagi misolni ko'rib chiqing:


```python
a = np.arange(3)
b = np.arange(3)[:, np.newaxis]
print(a)
print(b)
```

```python
a + b
```

# Radioeshittirish qoidalari


NumPy-da eshittirish ikki massiv o'rtasidagi o'zaro ta'sirni aniqlash uchun qat'iy qoidalar to'plamiga amal qiladi.:


1-qoida: Agar ikkita massiv o'lchamlari soni bo'yicha farq qilsa, o'lchamlari kamroq bo'lgan massivning shakli oldingi (chap) tomonida birlar bilan to'ldiriladi

2-qoida: Agar ikkita massivning shakli hech qanday o‘lchamda mos kelmasa, bu o‘lchamdagi shakli 1 ga teng bo‘lgan massiv boshqa shaklga mos kelishi uchun cho‘ziladi.


 3-qoida: Agar biron-bir o'lchamda o'lchamlar mos kelmasa va hech biri 1 ga teng bo'lmasa, xatolik ko'tariladi.


Ushbu qoidalarni aniqroq qilish uchun keling, bir nechta misollarni batafsil ko'rib chiqaylik.

## Ikki o‘lchovli funksiyaning grafigini tuzish.
## Eshittirish juda foydali bo'lgan joydan biri ikki o'lchovli funktsiyalarga asoslangan tasvirlarni ko'rsatishdir. Agar biz z = f(x, y) funksiyasini aniqlamoqchi bo'lsak, tarmoq bo'ylab funksiyani hisoblash uchun translyatsiyadan foydalanish mumkin.:


```python
x = np.linspace(0, 5, 50)
y = np.linspace(0, 5, 50)[:, np.newaxis]
z = np.sin(x) ** 10 + np.cos(10 + y * x) * np.cos(x)
```

```python
%matplotlib inline 
import matplotlib.pyplot as plt
plt.imshow(z, origin='lower', extent=[0, 5, 0, 5],
                   cmap='viridis') 
plt.colorbar();
```
